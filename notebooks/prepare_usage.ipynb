Add splice site type to df, based on strand and partners information:  
  - If strand=='+' and Partners > Site --> Donor
  - If strand=='+' and Partners < Site --> Acceptor
  - If strand=='-' and Partners > Site --> Acceptor
  - If strand=='-' and Partners < Site --> Donor

```python
def annotate_splice_site_type(df):
    """
    A function to annotate splice site type based on the strand and partners column.
    Partners column contains a string e.g. '{10: 100, 50: 0}' - this is a dictionary with partner positions (10, 50) as keys, and number of reads for given partner as values (100, 0).
    Add splice site type to df, based on strand and partners information:  
    - If strand=='+' and Partners > Site --> Donor
    - If strand=='+' and Partners < Site --> Acceptor
    - If strand=='-' and Partners > Site --> Acceptor
    - If strand=='-' and Partners < Site --> Donor
    """
    splice_site_types = []
    for row in df.itertuples():
        region = row.Region
        site = row.Site
        strand = row.Strand
        partners = eval(row.Partners)
        if strand == "+":
            donor_partners = [p for p in partners if p > site]
            acceptor_partners = [p for p in partners if p < site]
        else:
            donor_partners = [p for p in partners if p < site]
            acceptor_partners = [p for p in partners if p > site]
        
        if len(donor_partners) > 0 and len(acceptor_partners) > 0:
            splice_site_types.append("Both")
        elif len(donor_partners) > 0:
            splice_site_types.append("Donor")
        elif len(acceptor_partners) > 0:
            splice_site_types.append("Acceptor")
        else:
            splice_site_types.append("None")
    df['Splice_Site_Type'] = splice_site_types
    return df
```

Adjust splice site position:  
  - If strand=='+', Acceptor Site should be as is  
  - If strand=='+', Donor site should be -2  
  - If strand=='-', Acceptor Site should be -2  
  - If strand=='-', Donor site should be as is  

```python
def adjust_splice_site_type(df):
    """
    A function to adjust splice site positions.
    - If strand=='+', Acceptor site should be as is  
    - If strand=='+', Donor site should be -2  
    - If strand=='-', Acceptor Site should be -2  
    - If strand=='-', Donor site should be as is  
    """
    df['Site'] = df.apply(lambda row: row.Site - 2 if (row.Strand == '+' and row.Splice_Site_Type == 'Donor') or (row.Strand == '-' and row.Splice_Site_Type == 'Acceptor') else row.Site, axis=1)
    return df
```

Write a function to convert to parquet file

```python
import json
from zipfile import Path
def convert_usage_to_parquet(
    df: pd.DataFrame,
    class_labels: dict[str, int] = None,
    condition_labels: dict[str, int] | None = None,
    output_path: str | Path | None = None,
    compression: str = "snappy"
) -> None:
    """
    A function to convert usage dataframe to parquet format.
    Convert csv to parquet with columns: position strand label condition_idx sse alpha beta
    """
    df = df.copy()[['Tissue', 'Timepoint', 'Region', 'Site', 'Strand', 'Splice_Site_Type', 'SSE', 'alpha_count', 'beta1_count', 'beta2_count']].drop_duplicates()
    # Sum beta1 and beta2 counts into one beta count
    df['beta_count'] = df['beta1_count'] + df['beta2_count']
    df = df.drop(columns=['beta1_count', 'beta2_count'])
    # Convert to correct dtypes
    df["Tissue"] = df["Tissue"].astype(str)
    df["Timepoint"] = df["Timepoint"].astype(int)
    df["Region"] = df["Region"].astype(str)
    df["Site"] = df["Site"].astype(int)
    df["Strand"] = df["Strand"].astype(str)
    df["SSE"] = df["SSE"].astype(float)
    df["alpha_count"] = df["alpha_count"].astype(int)
    df["beta_count"] = df["beta_count"].astype(int)
    # For Both, make separate entries for Donor and Acceptor
    both_mask = df['Splice_Site_Type'] == 'Both'
    df_both = df[both_mask]
    df_both_donor = df_both.copy()
    df_both_donor['Splice_Site_Type'] = 'Donor'
    df_both_acceptor = df_both.copy()
    df_both_acceptor['Splice_Site_Type'] = 'Acceptor'
    df = pd.concat([df[~both_mask], df_both_donor, df_both_acceptor], ignore_index=True)
    # Map Splice_Site_Type to class labels
    df['Splice_Site_Type'] = df['Splice_Site_Type']+df['Strand']
    # None+, None- should be mapped to None
    df.loc[df['Splice_Site_Type'].str.startswith('None'), 'Splice_Site_Type'] = 'None'
    # Map to class labels
    if class_labels is None:
        class_labels = {'Donor+': 0, 'Acceptor+': 1, 'Donor-': 2, 'Acceptor-': 3, 'None': 4}
    df['Label'] = df['Splice_Site_Type'].map(class_labels)
    # Add class mapping to metadata json
    metadata = {'class_labels': class_labels}
    # Drop Splice_Site_Type and Strand columns
    df = df.drop(columns=['Splice_Site_Type', 'Strand'])
    # Keep track of condition idx (tissue and timepoint)
    df['Condition_Idx'] = df['Tissue'] + "_" + df['Timepoint'].astype(str)
    if condition_labels is None:
        condition_labels = {cond: idx for idx, cond in enumerate(df['Condition_Idx'].unique())}
    df['Condition_Idx'] = df['Condition_Idx'].map(condition_labels)
    # Add cond mapping to metadata json
    metadata['condition_labels'] = condition_labels
    # Drop Tissue and Timepoint columns
    df = df.drop(columns=['Tissue', 'Timepoint'])
    # Rename columns to Chromosome, Position, Condition
    df = df.rename(columns={'Region': 'Chromosome', 'Site': 'Position', "alpha_count": "Alpha", "beta_count": "Beta", 'Condition_Idx': 'Condition'})
    # Save parquet file and metadata json
    if output_path is not None:
        df.to_parquet(output_path, index=False, compression=compression)
        with open(output_path.replace(".parquet", ".json"), 'w') as f:
            json.dump(metadata, f)
    if output_path is None:
        return df
```

### Splice annotations

```python
import argparse
import sys
import time
from pathlib import Path

def convert_gtf_to_parquet(
    input_path: str | Path,
    output_path: str | Path | None = None,
    compression: str = "snappy",
) -> None:
    """Convert GTF file to Parquet format.

    Args:
        input_path: Path to input GTF/GFF file
        output_path: Path to output Parquet file
        compression: Parquet compression codec ('snappy', 'gzip', 'zstd', 'none')
    """
    import pandas as pd
    import pyranges

    input_path = Path(input_path)
    if output_path is not None:
        output_path = Path(output_path)

    if not input_path.exists():
        raise FileNotFoundError(f"Input file not found: {input_path}")

    print(f"Reading GTF file: {input_path}")
    print("This may take a few minutes for large files...")

    start = time.perf_counter()
    gtf = pyranges.read_gtf(str(input_path))
    read_time = time.perf_counter() - start
    print(f"  Read GTF in {read_time:.1f}s")

    # Convert to DataFrame
    df = gtf.df

    print(f"  Total features: {len(df):,}")
    print(f"  Feature types: {df['Feature'].value_counts().to_dict()}")

    # Optimize column types for smaller file and faster loading
    print("Optimizing column types...")

    # Convert string columns with low cardinality to categorical
    categorical_columns = [
        'Chromosome', 'Feature', 'Strand', 'gene_type', 'gene_biotype',
        'transcript_type', 'transcript_biotype', 'tag', 'Source'
    ]
    for col in categorical_columns:
        if col in df.columns:
            df[col] = df[col].astype('category')

    # Ensure integer columns are proper ints
    int_columns = ['Start', 'End', 'exon_number', 'level']
    for col in int_columns:
        if col in df.columns:
            # Handle NaN values by converting to nullable Int64
            if df[col].isna().any():
                df[col] = df[col].astype('Int64')
            else:
                df[col] = df[col].astype('int64')

    # Write to Parquet if output_path is not None
    if output_path is not None:
        print(f"Writing Parquet file: {output_path}")
        start = time.perf_counter()

        compression_arg = None if compression == 'none' else compression
        df.to_parquet(output_path, compression=compression_arg, index=False)

        write_time = time.perf_counter() - start
        print(f"  Wrote Parquet in {write_time:.1f}s")

        # Report file sizes
        input_size_mb = input_path.stat().st_size / (1024 * 1024)
        output_size_mb = output_path.stat().st_size / (1024 * 1024)

        print(f"\nConversion complete!")
        print(f"  Input size:  {input_size_mb:.1f} MB")
        print(f"  Output size: {output_size_mb:.1f} MB")
        print(f"  Compression ratio: {input_size_mb / output_size_mb:.1f}x")
        
    else:
        return df
```
